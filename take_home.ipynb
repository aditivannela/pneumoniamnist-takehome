```python
val_dataset = PneumoniaMNIST(split="val", download=True)
len(val_dataset)
```

```python
test_dataset = PneumoniaMNIST(split="test", download=True)
len(test_dataset)
```

```python
import numpy as np
import pandas as pd
import hashlib

# Put all three datasets together
datasets = {
    "train": train_dataset,
    "val": val_dataset,
    "test": test_dataset
}

# This will hold one dictionary/record per image
records = []

# Go through each split
for split_name, dataset in datasets.items():

    # Go through each image in that split
    for image_id, (image, label) in enumerate(dataset):

        # Convert PIL image to NumPy array
        image_array = np.array(image)

        # Create one metadata record for this image
        record = {
            "image_id": f"{split_name}_{image_id}",
            "split": split_name,
            "label": int(label[0]),
            "width": image_array.shape[1],
            "height": image_array.shape[0],
            "mean_intensity": np.mean(image_array),
            "std_intensity": np.std(image_array),
            "min_intensity": np.min(image_array),
            "max_intensity": np.max(image_array),
            "image_hash": hashlib.sha256(image_array.tobytes()).hexdigest()
        }

        # Add this image's record to our list
        records.append(record)

# Convert all records into a table
metadata_df = pd.DataFrame(records)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    recall_score,
    confusion_matrix,
    roc_auc_score,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
import numpy as np


# Convert each dataset into X (images) and y (labels)
def prepare_data(dataset):
    X = []
    y = []

    for image, label in dataset:
        image_array = np.array(image)

        # Flatten 28x28 image into 784 features
        flattened_image = image_array.flatten()

        # Scale pixel values from 0-255 to 0-1
        flattened_image = flattened_image / 255.0

        X.append(flattened_image)
        y.append(int(label[0]))

    return np.array(X), np.array(y)


# Prepare train, validation, and test data
X_train, y_train = prepare_data(train_dataset)
X_val, y_val = prepare_data(val_dataset)
X_test, y_test = prepare_data(test_dataset)

print("Train:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Test:", X_test.shape, y_test.shape)
```

```python
model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)
```

```python
val_predictions = model.predict(X_val)
val_probabilities = model.predict_proba(X_val)[:, 1]

val_accuracy = accuracy_score(y_val, val_predictions)
val_sensitivity = recall_score(y_val, val_predictions)

tn, fp, fn, tp = confusion_matrix(y_val, val_predictions).ravel()
val_specificity = tn / (tn + fp)

val_auc = roc_auc_score(y_val, val_probabilities)

print("Validation Accuracy:", round(val_accuracy, 3))
print("Validation Sensitivity:", round(val_sensitivity, 3))
print("Validation Specificity:", round(val_specificity, 3))
print("Validation ROC AUC:", round(val_auc, 3))
```

```python
test_predictions = model.predict(X_test)
test_probabilities = model.predict_proba(X_test)[:, 1]

test_accuracy = accuracy_score(y_test, test_predictions)
test_sensitivity = recall_score(y_test, test_predictions)

tn, fp, fn, tp = confusion_matrix(y_test, test_predictions).ravel()
test_specificity = tn / (tn + fp)

test_auc = roc_auc_score(y_test, test_probabilities)

print("Test Accuracy:", round(test_accuracy, 3))
print("Test Sensitivity:", round(test_sensitivity, 3))
print("Test Specificity:", round(test_specificity, 3))
print("Test ROC AUC:", round(test_auc, 3))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    test_predictions,
    display_labels=["Normal", "Pneumonia"]
)

plt.title("Logistic Regression - Test Set")
plt.show()
```
